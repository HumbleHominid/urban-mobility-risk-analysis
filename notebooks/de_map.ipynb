# Map of All Reported Traffic Accidents in Germany (Latest Year) and Frankfurt am Main (All Years)

```python
%load_ext autoreload
%autoreload 2
import matplotlib.pyplot as plt

from urban_mobility import fetch_data as fd
from urban_mobility import plotting as pl

fd.fetch_traffic_data()
```

```python
YEARS = fd.DATA_YEARS
df = fd.get_df(YEARS[-1])
df.head()
```

```python
fig, ax = plt.subplots(figsize=(15, 15))
pl.silhouette(df, by="ULAND", ax=ax)
pl.save(fig, f"de-{YEARS[-1]}", dpi=360, transparent=True)
```

```python
fig, ax = plt.subplots(figsize=(15, 15))
pl.silhouette(df, ax=ax)
pl.save(fig, f"de-{YEARS[-1]}-bw", dpi=360, transparent=True)
```

```python
df = fd.get_city_accidents("Frankfurt am Main", YEARS).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(15, 15))
pl.silhouette(df, size=1, alpha=0.8, ax=ax)
pl.save(fig, f"ffm-{YEARS[0]}-{YEARS[-1]}-bw", dpi=360, transparent=True)
```
