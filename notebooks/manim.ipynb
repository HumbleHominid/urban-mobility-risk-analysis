# Animating with Manim
Animations for the presentation, rendered to `output.media` (see `configs/config.toml`). Rendering is slow; the Germany scene draws only a sample of the accidents (`SAMPLE_FRAC`).

```python
import os

import pandas as pd
from manim import *  # type: ignore

from urban_mobility import fetch_data as fd
from urban_mobility.utils import cached_df, load_config, repo_root, resolve_path

# The %%manim magic needs the output under the working directory
os.chdir(repo_root())
cfg = load_config()
config.media_dir = str(resolve_path(cfg.output.media))
config.video_dir = "{media_dir}/videos/{quality}"
YEARS = fd.DATA_YEARS
```

## Lorenz Curve
How a Lorenz curve and the area behind the Gini index are built, on the example $L(x) = x^3$.

```python
%%manim -qm PlotLorenz

class PlotLorenz(Scene):
    def construct(self):
        plot_axes = Axes(
            x_range=[0, 1, 0.1],
            y_range=[0, 1, 0.1],
            x_length=5,
            y_length=5,
            axis_config={
                "numbers_to_include": np.arange(0, 1 + 0.2, 0.2),
                "font_size": 24,
            },
            tips=False,
        )

        x_label = plot_axes.get_x_axis_label(Tex("Proportion of X"), edge=DOWN, direction=DOWN)
        y_label = plot_axes.get_y_axis_label(Tex("Proportion of Y").rotate(90 * DEGREES), edge=LEFT, direction=LEFT, buff=0.4)
        plot_labels = VGroup(x_label, y_label)

        plot = plot_axes.plot(lambda x: x**3, color=WHITE)
        equality_line = plot_axes.plot(lambda x: x)

        extras = VGroup()
        extras += plot_axes.get_horizontal_line(plot_axes.c2p(1, 1, 0), color=BLUE)
        extras += plot_axes.get_vertical_line(plot_axes.c2p(1, 1, 0), color=BLUE)
        extras += DashedLine(start=plot_axes.c2p(0, 0, 0), end=plot_axes.c2p(1, 1, 0), color=RED)
        dot = Dot(point=plot_axes.c2p(1, 1, 0), color=YELLOW)
        
        title = Title(
            r"Lorenz Curve: $L(x) = x^3$",
            include_underline=False,
            font_size=40,
        )

        area = plot_axes.get_area(plot, [0, 1], bounded_graph=equality_line, color=GREY, opacity=0.5)
        
        self.play(Write(title))
        self.wait(3)
        self.play(Create(plot_axes), Create(plot_labels), lag_ratio=0, run_time=2)
        self.wait()
        self.play(Create(extras), lag_ratio=0.2, run_time=2)
        self.play(Create(dot))
        self.wait(5)
        self.play(Create(plot), run_time=10)
        self.wait()
        self.play(Create(area), run_time=5)
        self.wait(10)
```

## Accidents in Germany
A random sample of the latest year's accidents, appearing month by month and colored by state.

```python
%%manim --disable_caching -qm PlotAccidents

SAMPLE_FRAC = 0.1  # share of the accidents to draw; 1 draws all (~270k dots, very slow)
year = YEARS[-1]
df = fd.get_df(year).sample(frac=SAMPLE_FRAC, random_state=42).sort_values("UMONAT")
print(f"Sampling {len(df):,} accidents ({SAMPLE_FRAC:.0%}) of {year} for plotting.")

class PlotAccidents(Scene):
    def construct(self):
        x_min, x_max = int(df["LINREFX"].min()), int(df["LINREFX"].max())
        y_min, y_max = int(df["LINREFY"].min()), int(df["LINREFY"].max())
        plot_axes = Axes(
            x_range=[x_min, x_max, 100000],
            y_range=[y_min, y_max, 100000],
            x_length=x_max - x_min,
            y_length=y_max - y_min,
        )
        plot_axes.scale_to_fit_height(config.frame_height * 0.8)
        plot_axes.to_edge(DOWN)

        colors = color_gradient([BLUE, YELLOW], 16)  # one per state (ULAND 01-16)
        dot_dict = {month: VGroup() for month in sorted(df["UMONAT"].unique())}
        for row in df.itertuples():
            dot_dict[row.UMONAT] += Dot(
                point=plot_axes.c2p(row.LINREFX, row.LINREFY, 0),
                color=colors[int(row.ULAND) - 1],
                radius=0.01,
                fill_opacity=0.1,
            )

        title = Title(
            f"Accidents with Personal Injury in {year}",
            include_underline=False,
            font_size=40,
        )
        self.play(Write(title))
        self.wait()
        self.play(AnimationGroup(*[Create(dot_group) for dot_group in dot_dict.values()], lag_ratio=1))
        self.wait(10)
```

## Accidents in Frankfurt am Main
All Frankfurt accidents, appearing year by year and colored by participant: pedestrian (red), cyclist (green), motorcycle (orange), goods vehicle (purple), other (blue).

```python
%%manim --disable_caching -qm PlotFFM

df = fd.get_city_accidents("Frankfurt am Main", YEARS).sort_values(["UJAHR", "UMONAT", "USTUNDE"])
print(f"Rendering {len(df):,} accidents in Frankfurt am Main from {YEARS[0]} to {YEARS[-1]}.")

def participant_color(row):
    for col, color in [("IstFuss", RED), ("IstRad", GREEN), ("IstKrad", ORANGE), ("IstGkfz", PURPLE)]:
        if getattr(row, col) == 1:
            return color
    return BLUE

class PlotFFM(Scene):
    def construct(self):
        x_min, x_max = int(df["LINREFX"].min()), int(df["LINREFX"].max())
        y_min, y_max = int(df["LINREFY"].min()), int(df["LINREFY"].max())
        plot_axes = Axes(
            x_range=[x_min, x_max, 10000],
            y_range=[y_min, y_max, 10000],
            x_length=x_max - x_min,
            y_length=y_max - y_min,
            tips=False,
        )
        plot_axes.scale_to_fit_height(config.frame_height * 0.8)
        plot_axes.to_edge(DOWN)

        dots = VGroup()
        for year in YEARS:
            dots += VGroup(*[
                Dot(
                    point=plot_axes.c2p(row.LINREFX, row.LINREFY, 0),
                    color=participant_color(row),
                    radius=0.01,
                    fill_opacity=0.4,
                )
                for row in df[df["UJAHR"] == year].itertuples()
            ])

        title = Title(
            f"Accidents in FFM in {YEARS[0]}-{YEARS[-1]}",
            include_underline=False,
            font_size=40,
        )
        self.play(Write(title))
        self.wait()
        self.play(AnimationGroup(*[Create(dot_group) for dot_group in dots], lag_ratio=1))
        self.wait(5)
```

## Gini Index by State
Gini index of cyclist accidents per state and year, the same data as the state chart in `gini.ipynb` (shared cache). States without data in a year (see `metrics.STATES`) have no line segment there; Berlin, Hamburg and Bremen (fewer than 3 cities) are left out.

```python
%%manim -qm PlotGini

col = "IstRad"
df_gini = cached_df(
    f"gini/by_state_{col}_{YEARS[0]}-{YEARS[-1]}",
    lambda: gini_by_state(
        fd.get_city_aggregate(
            YEARS,
            ["Community_key", "UJAHR"],
            {"inj_light": "sum", "inj_serious": "sum", "inj_fatal": "sum", col: "sum", "ULAND": "first"},
        ),
        col,
    ),
)
by_land = df_gini.pivot(index="Year", columns="Land", values="Gini")
color_LUT = dict(zip(by_land.columns, color_gradient([YELLOW, BLUE], len(by_land.columns))))
y_lo = np.floor(df_gini["Gini"].min() / 0.04) * 0.04
y_hi = np.ceil(df_gini["Gini"].max() / 0.04) * 0.04

class PlotGini(Scene):
    def construct(self):
        plot_axes = Axes(
            x_range=[YEARS[0] - 1, YEARS[-1] + 1, 1],
            y_range=[y_lo, y_hi, 0.02],
            x_length=8,
            y_length=3,
            x_axis_config={
                "numbers_to_include": list(YEARS),
                "font_size": 20,
                "decimal_number_config": {"num_decimal_places": 0, "group_with_commas": False},
            },
            y_axis_config={"numbers_to_include": np.arange(y_lo, y_hi + 0.01, 0.04), "font_size": 20},
            tips=False,
        )
        plot_axes.scale_to_fit_width(config.frame_width * 0.75)
        plot_axes.to_edge(DOWN + 1)
        x_label = plot_axes.get_x_axis_label(Tex("Year"), edge=DOWN, direction=DOWN)
        y_label = plot_axes.get_y_axis_label(Tex("Gini-Index").rotate(90 * DEGREES), edge=LEFT, direction=LEFT, buff=0.4)
        title = Title(
            r"Gini-Index by State per Year",
            include_underline=False,
            font_size=40,
        )
        self.play(Create(title))
        self.wait()
        self.play(Create(plot_axes), Create(VGroup(x_label, y_label)))
        for prev, yr in zip(YEARS[:-1], YEARS[1:]):
            pair = by_land.loc[[prev, yr]].dropna(axis=1)  # states with data in both years
            lines = VGroup(*[
                plot_axes.plot_line_graph(
                    x_values=[prev, yr],
                    y_values=pair[land].tolist(),
                    add_vertex_dots=False,
                    line_color=color_LUT[land],
                    stroke_width=2,
                )
                for land in pair.columns
            ])
            self.play(Create(lines), run_time=2, lag_ratio=0)
            self.wait(1)
        self.wait(9)
```
